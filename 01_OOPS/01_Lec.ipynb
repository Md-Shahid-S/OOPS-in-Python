```python
'''
1. What is a Class and an Object?

The Mental Model:
Imagine you are building a fleet of houses.

    The Class is the architectural blueprint. It defines that a house has doors, windows, and a roof, 
    but it isn’t a physical thing. You can't live in a blueprint.

    The Object is the actual, physical house built using that blueprint. It takes up physical space (memory), 
    and you can paint its specific walls or open its specific doors (state and behavior).

In Python, a Class is a custom data type—a template that dictates what data and behaviors an entity should have. 
An Object (also called an Instance) is a concrete realization of that class residing in your computer's memory.

'''
```

```python
''' 
2 & 3. Defining a Class & The __init__ Method

To define a class, we use the class keyword followed by the name in PascalCase (capitalizing each word).

The __init__ method is Python's initializer (often casually called the constructor). It runs automatically the exact moment an object is created.

    self: This is a mandatory first parameter. It acts as a nametag that points to the specific object being created. 
    It tells Python, "attach these variables to me, not to the blueprint."

    Instance Attributes: Variables attached to self. They belong only to that specific object.

'''

class CloudServer:
    """Blueprint for a virtual private server."""
    
    # The __init__ method initializes the object's state
    def __init__(self, hostname, os_type, ram_gb):
        # These are instance attributes
        self.hostname = hostname      
        self.os_type = os_type        
        self.ram_gb = ram_gb          
        self.is_running = False       

# Creating the object (Instantiation) - see next section for details
server1 = CloudServer("web-prod-01", "Ubuntu 22.04", 16)
print(server1.hostname)
```

```python
''' 
Line-by-line breakdown:

    class CloudServer:: We define a new custom blueprint called CloudServer.

    def __init__(self, ...):: We define the setup process. When a server is built, it requires a hostname, OS, and RAM amount.

    self.hostname = hostname: We take the hostname passed by the user and attach it to this specific server's memory (self).

    self.is_running = False: We set a default instance attribute. All new servers start powered off.

    server1 = CloudServer(...): We build the object and store it in the variable server1. Python automatically passes server1 as the self argument.

    print(server1.hostname): We access the data inside server1.
    
    '''
```

```python
''' 
4. Creating Objects and Accessing Methods

A class isn't just about data (attributes); it's also about actions (methods). 
Methods are simply functions defined inside a class that operate on the object's data.

'''

class CloudServer:
    def __init__(self, hostname, os_type):
        self.hostname = hostname
        self.os_type = os_type
        self.is_running = False

    def boot(self):
        """Method to start the server."""
        if not self.is_running:
            self.is_running = True
            print(f"[{self.hostname}] Booting {self.os_type}...")
        else:
            print(f"[{self.hostname}] Server is already running.")

    def run_command(self, command):
        """Method to execute a command if the server is on."""
        if self.is_running:
            print(f"[{self.hostname}] Executing: {command}")
        else:
            print(f"[{self.hostname}] ERROR: Cannot run command. Server is offline.")

# 1. Instantiation
db_server = CloudServer("db-01", "Debian")

# 2. Accessing methods and attributes
db_server.run_command("sudo apt update")  # Fails, server is off
db_server.boot()                          # Turns server on
db_server.run_command("sudo apt update")  # Succeeds
```

```python
''' 
Line-by-line breakdown:

    def boot(self):: We define an action. self is required so the method knows which server to boot.

    if not self.is_running:: The method checks its own internal state.

    db_server = CloudServer(...): We instantiate the object.

    db_server.run_command(...): We call a method using dot notation. Python silently translates this to 
    CloudServer.run_command(db_server, "sudo apt update").


'''
```

```python
''' 
5. Class vs. Instance Attributes

    Instance Attributes (self.attribute): Belong to one specific object. Changing it on server A does not affect server B.

    Class Attributes (Class.attribute): Belong to the class itself. They are shared across all objects of that class.

When to use which: Use instance attributes for data unique to the object (like an IP address). Use class attributes for configuration, 
constants, or tracking data that applies globally (like the company domain or total server count).

'''

class CloudServer:
    # Class Attributes (Shared by ALL servers)
    company_domain = "internal.corp"
    total_servers_deployed = 0

    def __init__(self, hostname):
        # Instance Attributes (Unique to THIS server)
        self.hostname = hostname
        
        # Accessing class attribute to update global count
        CloudServer.total_servers_deployed += 1

    def get_fqdn(self):
        """Returns Fully Qualified Domain Name"""
        # Accessing instance attribute + class attribute
        return f"{self.hostname}.{CloudServer.company_domain}"

# Create two independent servers
server1 = CloudServer("web-01")
server2 = CloudServer("api-01")

print(f"Server 1 FQDN: {server1.get_fqdn()}")
print(f"Server 2 FQDN: {server2.get_fqdn()}")
print(f"Total Servers Deployed: {CloudServer.total_servers_deployed}")
```
